```python
import seaborn as sns
```

```python
import matplotlib.pyplot as plt
```

```python
numeric_df = df.select_dtypes(include=['number'])

plt.figure(figsize=(8,6))
sns.heatmap(numeric_df.corr(), annot=True, cmap='coolwarm')
plt.show()
```

```python
sns.scatterplot(data=df,x='study_hours',y='exam_score')
plt.show()

'''
There is a clear positive relationship between study hours and exam scores.
Students studying for longer durations tend to achieve better marks, making study hours one of the strongest predictors of academic performance.
'''
```

```python
df.groupby('study_hours')['exam_score'].mean()
```

```python
df.groupby('placement_status')['exam_score'].mean()
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(data=df, x='placement_status', y='exam_score')
plt.title("Exam Score Distribution by Placement Status")
plt.xlabel("Placement Status")
plt.ylabel("Exam Score")
plt.show()

'''
Students who passed consistently achieved higher exam scores than those who failed, indicating that exam score is strongly associated with the final outcome.
'''
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df['study_hours'], bins=10, kde=True)
plt.title("Distribution of Study Hours")
plt.xlabel("Study Hours")
plt.ylabel("Number of Students")
plt.show()

'''
The distribution of study hours shows that the majority of students dedicate a moderate amount of time (around 4 - 8 hours) to studying, while only a few students study for very short or very long durations.
'''
```
